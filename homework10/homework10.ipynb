```python
# All actual values are strings
# Using strings is a problem since strings can't naturally calculate like numbers
```

```python
for col in df.columns:
    if col not in ["Planet", "Ring System?", "Global Magnetic Field?"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.dtypes
```

```python
# float64, int64, str
# 64-bit floating point(not int) number; 64-bit integers; word strings
# the dtypes is still an object
```

```python
def attach_units(old_col, unit, new_col):
    new_values = []
    for val in df[old_col]:
        new_values.append(val*unit)
    df[old_col] = new_values
    df.rename(columns={old_col: new_col}, inplace=True)
```

```python
attach_units("Diameter (km)", 1000 * u.m, "Diameter (m)")
```

```python
semi_major_axis_qty = ((df["Perihelion (10^6 km)"] + df["Aphelion (10^6 km)"]) / 2 * 10000000 * u.m)
df.insert(12, "Semi-major axis", pd.Series(dtype=object, index=df.index))
df["Semi-major axis"] = list(semi_major_axis_qty)
```

```python
attach_units("Semi-major axis", u.m, "Semi-major axis (m)")
```

```python
attach_units('Diameter (m)', (1*u.m).to(u.au)/u.m, "Diameter (AU)")
attach_units('Distance from Sun (10^6 km)', (100000*u.km).to(u.au), "Distance from Sun (AU)")
attach_units('Perihelion (10^6 km)', (100000*u.km).to(u.au), 'Perihelion (AU)')
attach_units('Aphelion (10^6 km)', (100000*u.km).to(u.au),'Aphelion (AU)')
attach_units("Semi-major axis (m)", (1*u.m).to(u.au)/u.m, "Semi-major axis (AU)")
```
