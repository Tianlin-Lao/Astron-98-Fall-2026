```python
arr = np.array([[2, 3, 5], [14, 6, 8], [11, 13, 17], [7, 10, 13]])
def prime1(arr):
    def isprime(x):
        if x < 2:
            return False
        n = 2
        while n * n <= x:
            if x % n == 0:
                return False
            n += 1
        return True
    mask = np.array([any(isprime(x) for x in row) for row in arr])
    return arr[mask]
print(prime1(arr))
```

```python
def checkboard():
    ans = np.zeros((8, 8))
    ans[1::2, 1::2] = 1
    ans[::2, ::2] = 1
    return ans
checkboard()
```

5.4

```python
def reverse_checkboard():
    ans = np.zeros((8, 8))
    ans[1::2, ::2] = 1
    ans[::2, 1::2] = 1
    return ans
reverse_checkboard()
```

```python
def expansion(arr, n):
    sep = ' ' * n
    return np.char.join(sep, arr)
universe = np.array(['galaxy', 'clusters'])
expansion(universe, 1)
```
