```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
```

```python
print(data.shape) # 2642 rows, 2 columns
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
print('magnitude', min(magnitude), max(magnitude)) # giant stars and dwarf stars
print('temperature', min(temperature), max(temperature)) # red stars and blue stars
```

```python
# mag 2.24 star is brightest, while mag 12.49 star is dimmest
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
fig, ax = plt.subplots(figsize = (8, 6))
plt.scatter(log_temperature, log_magnitude, label = 'stars')
ax.invert_xaxis()
ax.invert_yaxis()
plt.scatter(log_temperature[log_magnitude == min(log_magnitude)], log_magnitude[log_magnitude == min(log_magnitude)], color = 'black', marker = '*', s = 300, label = 'brightest')
plt.scatter(log_temperature[log_magnitude == max(log_magnitude)], log_magnitude[log_magnitude == max(log_magnitude)], color = 'purple', marker = '*', s = 300, label = 'dimmest')
plt.scatter(log_temperature[log_temperature == max(log_temperature)], log_magnitude[log_temperature == max(log_temperature)], color = 'green', marker = '*', s = 300, label = 'hottest')
plt.scatter(log_temperature[log_temperature == min(log_temperature)], log_magnitude[log_temperature == min(log_temperature)], color = 'orange', marker = '*', s = 300, label = 'coldest')
plt.legend()
plt.xlabel('Temperature (log K)')
plt.ylabel('Magnitude')
plt.title('Hertzsprung-Russel Diagram')
plt.show()
```
