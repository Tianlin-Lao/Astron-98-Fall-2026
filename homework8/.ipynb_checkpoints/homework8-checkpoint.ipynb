```python
import numpy as np
import matplotlib.pyplot as plt
```

Part 2

```python
def make_sine_wave(x, A, w):
    return A*np.sin(w*x)
```

```python
x = np.linspace(0, 2*np.pi, 1000)
```

```python
amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])
```

```python
fig, axes = plt.subplots(1, 3, figsize=(25, 5))
for i, (A, w) in enumerate(zip(amplitude, frequency)):
    y1 = make_sine_wave(x, A, 1)
    axes[0].plot(x, y1, label=f'A = {A}')
    y2 = make_sine_wave(x, 1, w)
    axes[1].plot(x, y2, label=f'ω = {w}')
    y3 = make_sine_wave(x, A, w)
    axes[2].plot(x, y3, label=f'A = {A}, ω = {w}')
axes[0].set_title('Varying Amplitude (ω = 1)', fontsize=18)
axes[0].set_xlabel('x', fontsize=16)
axes[0].set_ylabel('y', fontsize=16)
axes[0].legend(fontsize=14)
axes[1].set_title('Varying Frequency (A = 1)', fontsize=18)
axes[1].set_xlabel('x', fontsize=16)
axes[1].set_ylabel('y', fontsize=16)
axes[1].legend(fontsize=14)
axes[2].set_title('Varying Amplitude and Frequency', fontsize=18)
axes[2].set_xlabel('x', fontsize=16)
axes[2].set_ylabel('y', fontsize=16)
axes[2].legend(fontsize=14)
plt.show()
```
